You are provided with a dataset of SMS messages classified as spam and non-spam, ```spam.csv```.

**Goal:** Build a model that will classify a message as spam/non-spam using the Naive Bayes algorithm.

Please review the ```dataset.py``` and ```model.py``` modules.

Message classification accuracy should be greater than 95% for both the validation and test datasets.

````python
import pandas as pd
from dataset import Dataset
from model import Model
````

````python
df = pd.read_csv('spam.csv', encoding='latin-1')
print("Columns:", df.columns.tolist())
print("First few rows:")
print(df.head())
````

````python
if 'v1' in df.columns and 'v2' in df.columns:
    labels = df['v1'].values
    messages = df['v2'].values
elif 'label' in df.columns and 'message' in df.columns:
    labels = df['label'].values
    messages = df['message'].values
else:
    labels = df.iloc[:, 0].values
    messages = df.iloc[:, 1].values

print(f"Dataset size: {len(messages)}")
print(f"Label distribution: {pd.Series(labels).value_counts()}")
````

````python
dataset = Dataset(messages, labels)
dataset.split_dataset(val=0.1, test=0.1)

print(f"Train size: {len(dataset.train[0])}")
print(f"Validation size: {len(dataset.val[0])}")
print(f"Test size: {len(dataset.test[0])}")
````

````python
model = Model(alpha=1)
model.fit(dataset)

print(f"Vocabulary size: {model.Nvoc}")
print(f"Total spam words: {model.Nspam}")
print(f"Total ham words: {model.Nham}")
````

````python
test_messages = [
    "Free money! Click here now!",
    "Hey, how are you doing today?",
    "WINNER! You've won a free prize!",
    "Can we meet for lunch tomorrow?"
]

print("Testing inference")
for msg in test_messages:
    prediction = model.inference(msg)
    print(f"Message: '{msg}' - {prediction}")
````

````python
val_acc = model.validation()
print(f"Validation Accuracy: {val_acc:.4f}")
````

````python
test_acc = model.test()
print(f"Test Accuracy: {test_acc:.4f} ")
````
